# Experiment - 8
Write a Python program to create, access, update and delete dictionaries.

# Create a dictionary

```python

employees = {
    101: {
        "name": "Aman",
        "department": "HR",
        "salary": 55000
    },

    102: {
        "name": "Bhavneet",
        "department": "IT",
        "salary": 100000
    },

    103: {
        "name": "Chandan",
        "department": "Finance",
        "salary": 62000
    }
}

print("Employee records:")
print(employees)
```

# 2. Access elements

```python
print("\nACCESS EMPLOYEE DETAILS\n")

employee_id = 102

print("Employee ID :", employee_id)
print("Name        :", employees[employee_id]["name"])
print("Department  :", employees[employee_id]["department"])
print("Salary      :", employees[employee_id]["salary"])
```

# 3. Update values

```python
print("\nUPDATE EMPLOYEE DETAILS\n")

print("Before update:")
print(employees[101])

employees[101]["salary"] = 60000

employees[101]["department"] = "Data Science"

print("\nAfter update:")
print(employees[101])
```

# 4. Add a new entry

```python
print("\nADD NEW EMPLOYEE\n")

employees[104] = {
    "name": "Daman",
    "department": "Marketing",
    "salary": 51000
}

print("New employee added:")
print(employees[104])
```

# 5. Delete a record

```python
print("\nDELETE EMPLOYEE\n")

deleted_employee = employees.pop(103)

print("Deleted employee:")
print(deleted_employee)

print("\nRemaining employees:")
print(employees)
```

# 6. clear dictionary

```python
print("\nCLEAR DICTIONARY\n")

employees.clear()

print("After clear():")
print(employees)
```

# Write a program for looping through dictionaries

```python
employees = {
    101: "Aman",
    102: "Bhavneet",
    103: "Chandan",
    104: "Daman"
}
```

```python
print("Employee IDs:")
for key in employees.keys():
    print(key)

print("\nEmployee Names:")
for value in employees.values():
    print(value)

print("\nEmployee Details:")
for key, value in employees.items():
    print("ID:", key, "Name:", value)
```

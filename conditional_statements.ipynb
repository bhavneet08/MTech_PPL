# Write Python program to demonstrate the use of conditional statements

# 1). If statement

```python
age = 20
if age >= 18:
    print("Eligible to vote.")
```

# 2). Short hand if

```python
age = 19
if age > 18: print("Eligible to Vote.")
```

# 3). If-else statement

```python
num=7
if num % 2 == 0:
    print("Even number")
else:
    print("Odd number")
```

# 4). If-elif-else statement

```python
marks = int(input("Enter your marks: "))

if marks >= 90:
    print("Grade: A")
elif marks >= 75:
    print("Grade: B")
elif marks >= 60:
    print("Grade: C")
elif marks >= 40:
    print("Grade: D")
else:
    print("Grade: F")
    print("Result: Fail")
```

# 5). Nested-if-else statement

```python
a=100
b=68
c=205

if a > b :
    if a > c:
        print("A is the greatest")
    else:
        print("C is the greatest")
else:
    if b > c:
        print("B is the greatest")
    else:
        print("C is the greatest")
```

# 6). Conditional Expression

```python
num = 6

x = "WEEKEND!" if num > 5 else "Workday"

print(x)
```

# Experiment-6

Write a Python program to create, access, update and delete lists. Write a
program to develop a real life list.

#   1. Create a list

```python
print("\nCREATE A LIST\n")

marks = [78, 85, 92, 67, 88]

print("Student marks:", marks)
```

# 2. Acess the list

```python
print("\nACCESS ELEMENTS\n")

print("First subject marks :", marks[0])
print("Second subject marks:", marks[1])
print("Last subject marks  :", marks[-1])

print("First three marks   :", marks[0:3])
```

# 3.Update the list

```python
print("\nUPDATE ELEMENTS\n")
print("Before update:", marks)

marks[1] = 90

print("After updating second subject:", marks)

marks[3:5] = [75, 91]

print("After updating multiple marks:", marks)
```

# 4. Add elements

```python
print("\nADD ELEMENTS\n")

marks.append(84)

print("After append():", marks)

marks.insert(2, 95)

print("After insert():", marks)

marks.extend([76, 89])

print("After extend():", marks)
```

# 5.Delete elements

```python
print("\nDELETE ELEMENTS\n")

marks.remove(75)

print("After remove():", marks)

deleted_mark = marks.pop(2)

print("Deleted mark:", deleted_mark)
print("After pop():", marks)

del marks[0]

print("After del:", marks)
```

# 6. Other list operations

```python
print("\nOTHER LIST OPERATIONS\n")

print("Number of subjects:", len(marks))
print("Highest marks:", max(marks))
print("Lowest marks:", min(marks))
print("Total marks:", sum(marks))

marks.sort()

print("Sorted marks:", marks)
```

# 7. Clear the list

```python
print("\nDELETE ALL ELEMENTS\n")

marks.clear()

print("After clear():", marks)
```

# Real-life example: Store and manage the names and phone numbers of students.

```python
contacts = [
    ["Aman", "9876543210"],
    ["Simran", "9876543211"],
    ["Rahul", "9876543212"]
]

# DISPLAY CONTACTS
def display_contacts():
    print("\nStudent Contacts")

    if len(contacts) == 0:
        print("No contacts available.")
    else:
        for number, contact in enumerate(contacts, start=1):
            print(f"{number}. {contact[0]} - {contact[1]}")


# ADD CONTACT
def add_contact():
    name = input("Enter student name: ")
    phone = input("Enter phone number: ")

    contacts.append([name, phone])

    print("Contact added successfully.")


# UPDATE CONTACT
def update_contact():
    display_contacts()

    if len(contacts) == 0:
        return

    number = int(input("Enter contact number to update: "))

    if 1 <= number <= len(contacts):
        name = input("Enter new name: ")
        phone = input("Enter new phone number: ")

        contacts[number - 1] = [name, phone]

        print("Contact updated successfully.")
    else:
        print("Invalid contact number.")


# DELETE CONTACT
def delete_contact():
    display_contacts()

    if len(contacts) == 0:
        return

    number = int(input("Enter contact number to delete: "))

    if 1 <= number <= len(contacts):
        deleted = contacts.pop(number - 1)
        print("Deleted contact:", deleted)
    else:
        print("Invalid contact number.")


# MAIN PROGRAM
while True:
    print("\nSTUDENT CONTACT MANAGEMENT")
    print("1. Display contacts")
    print("2. Add contact")
    print("3. Update contact")
    print("4. Delete contact")
    print("5. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        display_contacts()
    elif choice == "2":
        add_contact()
    elif choice == "3":
        update_contact()
    elif choice == "4":
        delete_contact()
    elif choice == "5":
        print("Exiting program...")
        break
    else:
        print("Invalid choice.")
```

# Real-life example: Manage books in a library by adding, updating, displaying, and deleting book records.

```python
books = [
    "Python Programming",
    "Data Structures",
    "Machine Learning"
]

# DISPLAY BOOKS
def display_books():
    print("\nLibrary Books")

    if len(books) == 0:
        print("No books available.")
    else:
        for number, book in enumerate(books, start=1):
            print(f"{number}. {book}")


# ADD BOOK
def add_book():
    book = input("Enter book name: ")
    books.append(book)
    print("Book added successfully.")


# UPDATE BOOK
def update_book():
    display_books()

    if len(books) == 0:
        return

    number = int(input("Enter book number to update: "))

    if 1 <= number <= len(books):
        new_book = input("Enter new book name: ")
        books[number - 1] = new_book
        print("Book updated successfully.")
    else:
        print("Invalid book number.")


# DELETE BOOK
def delete_book():
    display_books()

    if len(books) == 0:
        return

    number = int(input("Enter book number to delete: "))

    if 1 <= number <= len(books):
        deleted = books.pop(number - 1)
        print("Deleted book:", deleted)
    else:
        print("Invalid book number.")


# MAIN PROGRAM
while True:
    print("\nLIBRARY BOOK MANAGEMENT")
    print("1. Display books")
    print("2. Add book")
    print("3. Update book")
    print("4. Delete book")
    print("5. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        display_books()
    elif choice == "2":
        add_book()
    elif choice == "3":
        update_book()
    elif choice == "4":
        delete_book()
    elif choice == "5":
        print("Exiting program...")
        break
    else:
        print("Invalid choice.")
```

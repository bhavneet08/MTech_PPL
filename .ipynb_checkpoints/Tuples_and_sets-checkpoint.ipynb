# Experiment - 7
Write a Python program to create, access, update and delete tuples.

# 1. create a tuple 

```python
print("\nCREATE A TUPLE\n")

student = (
    "Bhavneet Bhagat",
    "M.Tech AI",
    "NIT Jalandhar",
    2026
)

print("Student details:")
print(student)
```

# 2. Access tuple elements

```python
print("\nACCESS TUPLE ELEMENTS\n")

print("Name       :", student[0])
print("Course     :", student[1])
print("Institute  :", student[2])
print("Year       :", student[3])

print("\nLast element:", student[-1])
print("Course and Institute:", student[1:3])
```

# 3. update a tuple (cannot modified directly)

```python
print("\nUPDATE A TUPLE\n")

print("Original tuple:")
print(student)

# Tuples cannot be changed directly.
# Instead, create a new tuple.

student = (
    student[0],
    "M.Tech Machine Learning",
    student[2],
    student[3]
)

print("\nUpdated tuple:")
print(student)
```

# 4 . delete a tuple 

```python
print("\nDELETE THE TUPLE\n")

print("Tuple before deletion:")
print(student)

del student

print("The tuple has been deleted.")
```

# Write a Python program to create, access, update and delete sets.

```python
# REAL-LIFE EXAMPLE: LIBRARY BOOK MANAGEMENT USING SETS

# 1. CREATE SETS
print("\nCREATE SETS\n")

A = {"Python", "C++", "Java", "DBMS", "Operating System"}
B = {"Python", "Java", "Computer Networks", "DBMS"}

print("Set A =", A)
print("Set B =", B)


# 2. ACCESS SET ELEMENTS
print("\nACCESS SET ELEMENTS\n")

print("Books in Set A:")

for book in A:
    print(book, end=" ")

print()

book = "Python"

if book in A:
    print(f"{book} is available in Set A.")
else:
    print(f"{book} is not available in Set A.")


# 3. ADD ELEMENTS
print("\n\nADD ELEMENTS\n")

print("A before adding:", A)

A.add("Machine Learning")

print("A after adding Machine Learning:", A)

A.update({"Artificial Intelligence", "Data Science"})

print("A after adding Artificial Intelligence and Data Science:", A)


# 4. UPDATE A SET
print("\nUPDATE A SET\n")

print("A before update:", A)

A.remove("C++")
A.add("Cloud Computing")

print("A after replacing C++ with Cloud Computing:", A)


# 5. UNION
print("\nUNION\n")

print("A =", A)
print("B =", B)

union = A.union(B)

print("\nA ∪ B =", union)


# 6. INTERSECTION
print("\nINTERSECTION\n")

intersection = A.intersection(B)

print("A ∩ B =", intersection)


# 7. DELETE AN ELEMENT
print("\nDELETE AN ELEMENT\n")

print("A before deletion:", A)

A.remove("Machine Learning")

print("A after deleting Machine Learning:", A)


# 8. CLEAR THE SET
print("\nCLEAR THE SET\n")

A.clear()

print("A after clear():", A)


# 9. DELETE THE SET
print("\nDELETE THE SET\n")

del A

print("Set A has been deleted.")
```

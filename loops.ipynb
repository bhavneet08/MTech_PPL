write program to use of diffenert loops.

# 1. For loop

```python
print("1. FOR LOOP")
for i in range(1, 6):
    print("Number:", i)
```

# 2. While loop

```python
print("\n2. WHILE LOOP")
i = 1
while i <= 5:
    print("Number:", i)
    i += 1
```

# 3. Nested loop

```python
print("\n3. NESTED LOOP")
for i in range(1, 4):
    for j in range(1, 4):
        print(i, "*", j, "=", i * j)
```

# 4. Break statement

```python
print("\n4. BREAK STATEMENT")
for i in range(1, 10):
    if i == 5:
        break
    print(i)
```

# 5. Continue statement

```python
print("\n5. CONTINUE STATEMENT")
for i in range(1, 6):
    if i == 3:
        continue
    print(i)
```

# 6. Pass statement

```python
print("\n6. PASS STATEMENT")

for i in range(1, 6):
    if i == 3:
        pass
    print(i)
```
